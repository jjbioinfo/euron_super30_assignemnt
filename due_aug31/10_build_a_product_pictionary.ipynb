# Build a Product Dictionary

## Create the exact laptop dictionary provided in the assignment.

```python
laptop = {
    "brand": "Dell",
    "model": "XPS 15",
    "price": 120000,
    "ram": "16GB",
    "storage": "512GB SSD",
    "available": True,
}
```

## Display the complete product and selected values.

```python
print("\nOriginal laptop:", laptop)

print("\nBrand:", laptop["brand"])

print("\nModel:", laptop["model"])

print("\nPrice:", laptop["price"])
```

## Change the laptop price.

```python
laptop["price"] = 125000
print("Updated price:", laptop["price"])
```

## Add processor and GPU information.

```python
laptop["processor"] = "Intel Core i7"
laptop["gpu"] = "NVIDIA RTX 4060"

print("Processor:", laptop["processor"])

print("GPU:", laptop["gpu"])
```

## Upgrade the existing RAM value.

```python
laptop["ram"] = "32GB"
print("Updated RAM:", laptop["ram"])
```

## Remove available and save the value returned by pop().

```python
removed_availability = laptop.pop("available")

print("Removed availability value:", removed_availability)

print("Updated laptop:", laptop)
```

## Display the dictionary views requested in the assignment.

```python
print("All laptop keys:", laptop.keys())
print("All laptop values:", laptop.values())
print("All laptop items:", laptop.items())
```

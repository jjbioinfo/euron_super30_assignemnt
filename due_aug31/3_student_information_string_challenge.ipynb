# Student Information String Challenge


## Create the exact string provided in the assignment.


```python
student = "python programming for data science"
```

```python
print("Complete string:", student)

print("\nFirst character:", student[0])

print("\nLast character:", student[-1])

print("\nFirst 6 characters:", student[:6])

print("\nLast 7 characters:", student[-7:])

print("\nReversed string:", student[::-1])
```

```python
print("\nUppercase:", student.upper())

print("\nLowercase:", student.lower())

print("\nTitle case:", student.title())

print('\nNumber of times "a" appears:', student.count("a"))

print('\nPosition of "programming":', student.find("programming"))

print(
    '\nReplace "data science" with "artificial intelligence":',
    student.replace("data science", "artificial intelligence"),
)
print("\nIndividual words:", student.split())
```

# Tuple Challenge

## Create the exact tuple provided in the assignment.

```python
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")
```

## Display the tuple, its type, and the required indexing results.

```python
print("\nOriginal tuple:", technologies)

print("\nTuple type:", type(technologies))
print("\nFirst item:", technologies[0])
print("\nLast item:", technologies[-1])
```

## Slice from index 1 up to, but not including, index 4.

```python
tuple_slice = technologies[1:4]
print("Slice from index 1 to 3:", tuple_slice)
```

## Use tuple methods and len().

```python
python_count = technologies.count("Python")
cpp_index = technologies.index("C++")
tuple_length = len(technologies)

print('Number of times "Python" appears:', python_count)
print('Index of "C++":', cpp_index)
print("Tuple length:", tuple_length)
```

## Convert the immutable tuple into a mutable list.

```python
technology_list = list(technologies)
print("Converted list:", technology_list)
print("Converted collection type:", type(technology_list))
```

## Add Go to the mutable list.

```python
technology_list.append("Go")
print('List after adding "Go":', technology_list)
```

## Convert the updated list back into a tuple.

```python
updated_technologies = tuple(technology_list)
print("Updated tuple:", updated_technologies)
print("Updated collection type:", type(updated_technologies))
```

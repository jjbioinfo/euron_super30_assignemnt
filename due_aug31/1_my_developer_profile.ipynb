# Build Developer Profile Using Variables

```python
## Create Python variables to store the following details

#Your name
name = "Jibin John"

#Age
age = 40

#City
city = "New York"

#College/company name
college_or_company_name = "The Feinstein Institutes for Medical Research, Northwell Health"

#Current learning topic
current_learning_topic = "Production AI engineering with Python"

#Years of experience or study
years_of_experience_or_study = 10

#Whether you like Python (True/False)
likes_python = True

#Your expected salary after learning Python
expected_salary_after_learning_python = 150000.0
```

## Print the value.



```python
print("MY DEVELOPER PROFILE")

print("Name:", name)

print("Age:", age)

print("City:", city)

print("College/Company:", college_or_company_name)

print("Current learning topic:", current_learning_topic)

print("Years of experience or study:", years_of_experience_or_study)

print("Likes Python:", likes_python)

print("Expected salary after learning Python:", expected_salary_after_learning_python)
```

##  Print the data type of every variable

```python
print("\nDATA TYPES")
print("Type of name:", type(name))
print("Type of age:", type(age))
print("Type of city:", type(city))
print("Type of college_or_company_name:", type(college_or_company_name))
print("Type of current_learning_topic:", type(current_learning_topic))
print("Type of years_of_experience_or_study:", type(years_of_experience_or_study))
print("Type of likes_python:", type(likes_python))
print(
    "Type of expected_salary_after_learning_python:",
    type(expected_salary_after_learning_python),
)
```

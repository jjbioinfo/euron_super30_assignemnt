# Nested Dictionary Challenge

## Create the exact nested dictionary provided in the assignment.

```python
employee = {
    "name": "Amit",
    "department": "Engineering",
    "skills": {
        "language": "Python",
        "database": "PostgreSQL",
        "cloud": "AWS",
    },
    "salary": 80000,
}
```

## Access the required values directly, without using loops.

```python
print("Employee name:", employee["name"])

print("\nDepartment:", employee["department"])

print("\nComplete skills dictionary:", employee["skills"])

print("\nProgramming language:", employee["skills"]["language"])

print("\nDatabase:", employee["skills"]["database"])

print("\nCloud technology:", employee["skills"]["cloud"])
```

## Change the language value inside the nested skills dictionary.

```python
employee["skills"]["language"] = "Python + JavaScript"
print("Updated programming language:", employee["skills"]["language"])
```

## Change the top-level salary value.

```python
employee["salary"] = 90000
print("Updated salary:", employee["salary"])
```

## Add a new top-level experience key-value pair.

```python
employee["experience"] = 3
print("Experience in years:", employee["experience"])
```

## Add another skill inside the nested skills dictionary.

```python
employee["skills"]["framework"] = "Django"
print("Added framework skill:", employee["skills"]["framework"])
```

## Print the complete updated structures.

```python
print("Updated skills dictionary:", employee["skills"])
print("Complete updated employee dictionary:", employee)
```

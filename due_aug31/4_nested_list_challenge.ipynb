# Nested List Challenge

## Create the exact nested list provided in the assignment.


```python
students = [
    ["Rahul", 21, "Python"],
    ["Priya", 22, "Data Science"],
    ["Aman", 20, "Machine Learning"],
]
```

## Access the required records and values directly, without using loops.


```python
print("\nComplete list:", students)

print("\nRahul's name:", students[0][0])

print("\nPriya's age:", students[1][1])

print("\nAman's course:", students[2][2])

print("\nPriya's complete record:", students[1])
```

## Change Rahul's course from Python to AI.

```python
students[0][2] = "AI"
print("Rahul's updated record:", students[0])
```

## Add one student record manually.

```python
students.append(["Neha", 23, "Cloud Computing"])
```

## Print the updated nested list without a loop.

```python
print("Updated nested list:", students)
```

# Student Profile Using Dictionary

## Create the exact student dictionary provided in the assignment.

```python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "city": "Bangalore",
    "marks": 88,
}
```

## Print the complete dictionary and selected values.

```python
print("Original student dictionary:", student)

print("Student name:", student["name"])

print("Student course:", student["course"])
```

## Display all dictionary keys, values, and key-value pairs.

```python
print("All keys:", student.keys())
print("All values:", student.values())
print("All key-value pairs:", student.items())
```

## Use update() to change marks from 88 to 92.

```python
student.update({"marks": 92})
print("Updated marks:", student["marks"])
```

## Use update() to add email and phone entries.

```python
student.update(
    {
        "email": "rahul@example.com",
        "phone": "9876543210",
    }
)
print("Added email:", student["email"])
print("Added phone:", student["phone"])
```

## Remove city and save the value returned by pop().

```python
removed_city = student.pop("city")
print("Removed city:", removed_city)
```

## Retrieve name safely with get().

```python
retrieved_name = student.get("name")
print('Name retrieved with get("name"):', retrieved_name)
```

## Create a shallow copy of the updated dictionary.

```python
student_copy = student.copy()
print("Updated student dictionary:", student)
print("Copied student dictionary:", student_copy)
print("Copy is a separate dictionary object:", student_copy is not student)
```

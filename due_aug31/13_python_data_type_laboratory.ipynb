## Demonstrate basic type conversion

```python
string_to_integer = int("100")
print('\nint("100") ->', string_to_integer, type(string_to_integer), "Text becomes an integer.")
```

```python
string_to_float = float("45.67")
print(
    '\nfloat("45.67") ->',
    string_to_float,
    type(string_to_float),
    "Text becomes a floating-point number.",
)
```

```python
integer_to_string = str(500)
print("\nstr(500) ->", repr(integer_to_string), type(integer_to_string), "An integer becomes text.")
```

```python
integer_to_boolean = bool(1)
print(
    "\nbool(1) ->",
    integer_to_boolean,
    type(integer_to_boolean),
    "A non-zero integer becomes True.",
)
```

```python
tuple_to_list = list((1, 2, 3))
print(
    "\nlist((1, 2, 3)) ->",
    tuple_to_list,
    type(tuple_to_list),
    "A tuple becomes a mutable list.",
)
```

```python
list_to_tuple = tuple([1, 2, 3])
print(
    "\ntuple([1, 2, 3]) ->",
    list_to_tuple,
    type(list_to_tuple),
    "A list becomes an immutable tuple.",
)
```

```python

list_to_set = set([1, 2, 2, 3])

print(
    "\nset([1, 2, 2, 3]) ->",
    list_to_set,
    type(list_to_set),
    "A list becomes a set, and the duplicate 2 is removed.",
)
```

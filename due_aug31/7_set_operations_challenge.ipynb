# Set Operations Challenge.

##  Create the exact sets provided in the assignment.

```python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
java_students = {"Priya", "Karan", "Rohit", "Simran"}

# Print both sets.
print("Python students:", python_students)
print("Java students:", java_students)
```

```python
#Find students learning either Python or Java
either_language = python_students.union(java_students)

print("\nStudents learning Python or Java or both:", either_language)
```

```python
#Find students learning both
both_languages = python_students.intersection(java_students)

print("Students learning both Python and Java:", both_languages)
```

```python
#Find students learning only Python
only_python = python_students.difference(java_students)
print("Students learning only Python:", only_python)
```

```python
#Find students learning only Java
only_java = java_students.difference(python_students)
print("Students learning exactly one language:", only_java)
```

```python
#Find students belonging to exactly one group
exactly_one_language = python_students.symmetric_difference(java_students)

print("Students learning exactly one language:", exactly_one_language)
```

## Add one new student to the Python group.

```python
python_students.add("Arjun")
print('\nAfter add("Arjun") to Python:', python_students)
```

## Remove one existing student from the Python group.


```python
python_students.remove("Neha")
print('After remove("Neha") from Python:', python_students)
```

## discard() removes an existing item without returning an error.

```python
java_students.discard("Rohit")
print('After discard("Rohit") from Java:', java_students)
```

## discard() also remains safe when the requested item is absent.


```python
java_students.discard("Vikram")
print('After discard("Vikram") when absent:', java_students)
```

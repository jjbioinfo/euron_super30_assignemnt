# String Cleaning Challenge


## Create the exact string provided in the assignment.


```python
message = " Welcome To Python Programming Class "
```

## Apply each required string operation.


```python
#Remove extra spaces
cleaned_message = message.strip()

#Convert everything to lowercase
lowercase_message = cleaned_message.lower()

#Convert everything to uppercase
uppercase_message = cleaned_message.upper()

#Convert to title case
title_case_message = cleaned_message.title()


#Replace "Python" with "Advanced Python"
replaced_message = cleaned_message.replace("Python", "Advanced Python")

#Check whether the string starts with "Welcome"
starts_with_welcome = cleaned_message.startswith("Welcome")

#Check whether it ends with "Class"
ends_with_class = cleaned_message.endswith("Class")

#Count occurrences of "o"
number_of_lowercase_o = cleaned_message.count("o")

#Find the position of "Programming"
programming_position = cleaned_message.find("Programming")

#Split the sentence into words
words = cleaned_message.split()
```

## Print the original string and the result of every operation.


```python

print("\nOriginal message:", repr(message))
print("\nExtra spaces removed:", cleaned_message)
print("\nLowercase:", lowercase_message)
print("\nUppercase:", uppercase_message)
print("\nTitle case:", title_case_message)
print("\nPython replaced:", replaced_message)
print('\nStarts with "Welcome":', starts_with_welcome)
print('\nEnds with "Class":', ends_with_class)
print('\nOccurrences of lowercase "o":', number_of_lowercase_o)
print('\nPosition of "Programming":', programming_position)
print("\nSentence split into words:", words)
```

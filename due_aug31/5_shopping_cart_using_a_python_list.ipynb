# Shopping Cart Using a Python List


## Create the exact shopping cart provided in the assignment.

```python
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
```

## Display the original products and access the first and last products.

```python
print("Original cart:", cart)
print("\nFirst product:", cart[0])
print("\nLast product:", cart[-1])
```

## Add Webcam at the end of the cart.


```python
cart.append("Webcam")
print('After append("Webcam"):', cart)
```

## Insert USB Hub at index 2.

```python
cart.insert(2, "USB Hub")
print('After insert(2, "USB Hub"):', cart)
```

## Remove Mouse by its value.


```python
cart.remove("Mouse")
print('After remove("Mouse"):', cart)
```

## Remove and save the last product.


```python
removed_product = cart.pop()
print("Product removed by pop():", removed_product)
print("After pop():", cart)
```

## Find and count products in the current cart.


```python
monitor_index = cart.index("Monitor")
laptop_count = cart.count("Laptop")
print('Index of "Monitor":', monitor_index)
print('Occurrences of "Laptop":', laptop_count)
```

## Make an independent shallow copy before changing the original cart again.


```python
cart_copy = cart.copy()
print("Copied cart:", cart_copy)
```

## Reverse the cart in place.

```python
cart.reverse()
print("Reversed cart:", cart)
print("Copy remains unchanged:", cart_copy)
```

## Sort the reversed cart alphabetically in place.


```python
cart.sort()
print("Alphabetically sorted cart:", cart)
```

## Demonstrate extend() and clear() on a separate list without changing the cart.


```python
method_demo = cart_copy.copy()
method_demo.extend(["Microphone", "Speakers"])
print("extend() demo:", method_demo)

method_demo.clear()
print("clear() demo:", method_demo)
```

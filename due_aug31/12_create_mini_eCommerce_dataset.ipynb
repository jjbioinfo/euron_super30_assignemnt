# Create a Mini E-Commerce Dataset

## Create three products using a list of dictionaries.

```python
products = [
    {
        "name": "Laptop",
        "price": 70000,
        "brand": "Dell",
    },
    {
        "name": "Phone",
        "price": 40000,
        "brand": "Samsung",
    },
    {
        "name": "Tablet",
        "price": 30000,
        "brand": "Apple",
    },
]
```

## Access the dataset directly, without loops.

```python
print("All original products:", products)


print("First product:", products[0])

print("Second product's price:", products[1]["price"])

print("Third product's brand:", products[2]["brand"])
```

## Display the keys and values of one product dictionary.

```python
print("First product's keys:", products[0].keys())

print("First product's values:", products[0].values())
```

## Change the first product's price.

```python
products[0]["price"] = 75000
print("First product after price change:", products[0])
```

## Add a rating to the second product.

```python
products[1]["rating"] = 4.5
print("Second product after adding rating:", products[1])
```

## Add another complete product dictionary manually.

```python
products.append(
    {
        "name": "Smartwatch",
        "price": 20000,
        "brand": "Garmin",
    }
)

# Print the final dataset without using a loop.
print("Final e-commerce dataset:", products)
```

# Remove Duplicate Data Using Sets

## Create the exact list provided in the assignment.

```python
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]
```

## Remove duplicate numbers without using loops.

```python
print("Original number list:", numbers)

unique_numbers = set(numbers)
print("\nSet of unique numbers:", unique_numbers)

numbers_without_duplicates = list(unique_numbers)
print("\nUnique numbers converted back to a list:", numbers_without_duplicates)

print("\nNumber of original elements:", len(numbers))

print("\nNumber of unique elements:", len(unique_numbers))
print("Number of duplicate entries removed:", len(numbers) - len(unique_numbers))
```

## Create another example containing duplicate student names.

```python
student_names = ["Rahul", "Priya", "Aman", "Priya", "Rahul", "Neha", "Aman"]

print("\nOriginal student names:", student_names)

unique_student_names = set(student_names)

print("\nSet of unique student names:", unique_student_names)

student_names_without_duplicates = list(unique_student_names)
print("\nUnique student names converted back to a list:", student_names_without_duplicates)

print("\nNumber of original student-name entries:", len(student_names))

print("\nNumber of unique student names:", len(unique_student_names))
```

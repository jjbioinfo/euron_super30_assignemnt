# Python for Loop Fundamentals

## 1. Print Numbers from 1 to 100

```python
for number in range(1, 101):
    print(number, end=" ")
```

## 2. Print Even Numbers from 1 to 100

```python
for number in range(2, 101, 2):
    print(number, end=" ")
```

## 3. Print Odd Numbers from 1 to 100

```python
for number in range(1, 101, 2):
    print(number,end=" ")
```

## 4. Take an integer n and print its multiplication table from 1 to 20.

Sample input: `5`

```python
n = int(input("Enter an integer: "))

for multiplier in range(1, 21):
    print(n, "x", multiplier, "=", n * multiplier)
```

## 5. Calculate the sum of numbers from 1 to n using a loop.

Sample input: `10`  
Expected result: `55`

```python
n = int(input("Enter an integer: "))
total = 0

for number in range(1, n + 1):
    total = total + number

print("Sum from 1 to", n, "=", total)
```

## 7. Print only the numbers divisible by 3.

```python
numbers = [12, 7, 9, 20, 33, 42, 8, 15]

for number in numbers:
    if number % 3 == 0:
        print(number)
```

## 8. Print Every Language and Its Length

```python
languages = ["Python", "Java", "C++", "JavaScript", "Go"]

for language in languages:
    print(language, "->", len(language))
```

## 9. Print Every Dictionary Key and Value

```python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Data Science",
    "city": "Bangalore"
}

for key, value in student.items():
    print(key, "->", value)
```

## 10. Count Vowels in a User-Provided String

Sample input: `education`  
Expected result: `5`

```python
text = input("Enter a string: ")
vowel_count = 0

for character in text.lower():
    if character in "aeiou":
        vowel_count = vowel_count + 1

print("Number of vowels:", vowel_count)
```

## 11. Reverse a string using a for loop without using [::-1] or reversed().

Sample input: `Python`  
Expected result: `nohtyP`

```python
text = input("Enter a string: ")
reversed_text = ""

for character in text:
    reversed_text = character + reversed_text

print("Reversed string:", reversed_text)
```

## 12. Find the Largest Number Without `max()`

```python
numbers = [45, 12, 89, 33, 67, 21,100]
largest = numbers[0]

for number in numbers[1:]:
    if number > largest:
        largest = number

print("Numbers:", numbers)
print("Largest number:", largest)
```

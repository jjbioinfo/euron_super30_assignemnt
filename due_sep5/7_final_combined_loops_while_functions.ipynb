## 1. Student Result Management System

Functions accept marks, calculate total and percentage, assign grade, determine pass/fail, and display the result.

Sample input: `3`, followed by `80`, `90`, and `70`.

```python
def accept_student_marks():
    """Accept student marks and return them as a list."""
    subject_count = int(input("Enter the number of subjects: "))
    marks = []

    for subject in range(1, subject_count + 1):
        mark = float(input(f"Enter mark for subject {subject}: "))
        marks.append(mark)

    return marks


def calculate_total(marks):
    """Return the total of all student marks."""
    total = 0

    for mark in marks:
        total = total + mark

    return total


def calculate_percentage(total, subject_count):
    """Return the percentage when each subject is marked out of 100."""
    if subject_count == 0:
        return 0

    return total / subject_count


def assign_grade(percentage):
    """Return a grade based on the supplied percentage."""
    if percentage >= 90:
        return "A+"
    if percentage >= 80:
        return "A"
    if percentage >= 70:
        return "B"
    if percentage >= 60:
        return "C"
    if percentage >= 50:
        return "D"

    return "F"


def determine_pass_fail(marks):
    """Return Fail if any subject mark is below 40; otherwise return Pass."""
    for mark in marks:
        if mark < 40:
            return "Fail"

    return "Pass"


def display_result(marks, total, percentage, grade, result):
    """Display the complete student result."""
    print("Marks:", marks)
    print("Total:", total)
    print(f"Percentage: {percentage:.2f}%")
    print("Grade:", grade)
    print("Result:", result)


def student_result_management_system():
    """Control the complete student result workflow."""
    marks = accept_student_marks()
    total = calculate_total(marks)
    percentage = calculate_percentage(total, len(marks))
    grade = assign_grade(percentage)
    result = determine_pass_fail(marks)
    display_result(marks, total, percentage, grade, result)


student_result_management_system()
```

## 2. Banking Application

Menu: Check balance, Deposit, Withdraw, Transaction history, and Exit.

Suggested test: check balance, deposit `500`, withdraw `300`, view history, and exit.

```python
def check_balance(balance):
    """Display the current bank balance."""
    print(f"Current balance: ₹{balance:.2f}")


def deposit(balance, history):
    """Accept a deposit and return the updated balance."""
    amount = float(input("Enter deposit amount: ₹"))

    if amount <= 0:
        print("Deposit amount must be greater than zero")
        return balance

    balance = balance + amount
    history.append(f"Deposited ₹{amount:.2f}")
    print(f"Deposit successful. Balance: ₹{balance:.2f}")
    return balance


def withdraw(balance, history):
    """Accept a withdrawal and return the updated balance."""
    amount = float(input("Enter withdrawal amount: ₹"))

    if amount <= 0:
        print("Withdrawal amount must be greater than zero")
    elif amount > balance:
        print("Insufficient balance")
    else:
        balance = balance - amount
        history.append(f"Withdrew ₹{amount:.2f}")
        print(f"Withdrawal successful. Balance: ₹{balance:.2f}")

    return balance


def show_transaction_history(history):
    """Display all completed banking transactions."""
    if len(history) == 0:
        print("No transactions available")
        return

    for transaction in history:
        print(transaction)


def banking_application():
    """Run the banking menu until Exit is selected."""
    balance = 10000.0
    history = []

    while True:
        choice = input("""
1. Check balance
2. Deposit
3. Withdraw
4. Transaction history
5. Exit
Enter choice (1-5): """).strip()

        if choice == "1":
            check_balance(balance)
        elif choice == "2":
            balance = deposit(balance, history)
        elif choice == "3":
            balance = withdraw(balance, history)
        elif choice == "4":
            show_transaction_history(history)
        elif choice == "5":
            print("Banking application closed")
            break
        else:
            print("Invalid choice")


banking_application()
```

## 3. Inventory Management

Products contain product name, price, and quantity. Functions add, display, search, update, and calculate total value.

```python
def add_product(products, product_name, price, quantity):
    """Add one product dictionary to the inventory list."""
    product = {
        "product_name": product_name,
        "price": price,
        "quantity": quantity
    }
    products.append(product)


def display_products(products):
    """Display every product in the inventory."""
    for product in products:
        print(product)


def search_product(products, product_name):
    """Return the product matching the supplied name or return None."""
    for product in products:
        if product["product_name"] == product_name:
            return product

    return None


def update_quantity(products, product_name, new_quantity):
    """Update a product quantity and return whether it was found."""
    product = search_product(products, product_name)

    if product is None:
        return False

    product["quantity"] = new_quantity
    return True


def calculate_total_inventory_value(products):
    """Return the total price of all inventory units."""
    total_value = 0

    for product in products:
        total_value = total_value + product["price"] * product["quantity"]

    return total_value


products = []
add_product(products, "Laptop", 55000, 2)
add_product(products, "Mouse", 700, 5)

print("Products:")
display_products(products)
print("Search Mouse:", search_product(products, "Mouse"))
print("Quantity updated:", update_quantity(products, "Mouse", 10))
print("Total inventory value:", calculate_total_inventory_value(products))
```

## 4. Quiz Application

Five Python questions are displayed one at a time. Answers are checked and the final score and percentage are shown.

```python
def run_quiz(questions):
    """Ask every quiz question and display the final score and percentage."""
    score = 0

    for question_data in questions:
        print("\n" + question_data["question"])
        answer = input("Your answer: ").strip().lower()

        if answer == question_data["answer"].lower():
            print("Correct")
            score = score + 1
        else:
            print("Incorrect. Correct answer:", question_data["answer"])

    percentage = score / len(questions) * 100
    print("\nFinal score:", score, "/", len(questions))
    print(f"Final percentage: {percentage:.2f}%")


questions = [
    {"question": "Which keyword defines a Python function?", "answer": "def"},
    {"question": "Which loop runs while a condition is true?", "answer": "while"},
    {"question": "Which data type stores key-value pairs?", "answer": "dictionary"},
    {"question": "What is the result of len([1, 2, 3])?", "answer": "3"},
    {"question": "Which keyword sends a value back from a function?", "answer": "return"}
]

run_quiz(questions)
```

## 5. Number Analysis Tool

Do not use `min()`, `max()`, or `sum()`.

```python
def analyze_numbers(numbers):
    """Return largest, smallest, total, average, and number-category counts."""
    if len(numbers) == 0:
        return None

    largest = numbers[0]
    smallest = numbers[0]
    total = 0
    even_count = 0
    odd_count = 0
    positive_count = 0
    negative_count = 0

    for number in numbers:
        total = total + number

        if number > largest:
            largest = number
        if number < smallest:
            smallest = number

        if number % 2 == 0:
            even_count = even_count + 1
        else:
            odd_count = odd_count + 1

        if number > 0:
            positive_count = positive_count + 1
        elif number < 0:
            negative_count = negative_count + 1

    average = total / len(numbers)

    return {
        "largest number": largest,
        "smallest number": smallest,
        "total": total,
        "average": average,
        "even count": even_count,
        "odd count": odd_count,
        "positive count": positive_count,
        "negative count": negative_count
    }


numbers = [12, -7, 0, 25, -4, 18, 9]
print("Numbers:", numbers)
print("Analysis:", analyze_numbers(numbers))
```

## 6. Employee Salary Analyzer

Functions determine total payroll, average salary, highest salary, lowest salary, and employees earning above average.

```python
def total_payroll(employees):
    """Return the total salary paid to all employees."""
    total = 0

    for salary in employees.values():
        total = total + salary

    return total


def average_salary(employees):
    """Return the average employee salary."""
    return total_payroll(employees) / len(employees)


def highest_salary(employees):
    """Return the employee with the highest salary."""
    highest_employee = None
    highest_value = None

    for employee, salary in employees.items():
        if highest_value is None or salary > highest_value:
            highest_employee = employee
            highest_value = salary

    return highest_employee, highest_value


def lowest_salary(employees):
    """Return the employee with the lowest salary."""
    lowest_employee = None
    lowest_value = None

    for employee, salary in employees.items():
        if lowest_value is None or salary < lowest_value:
            lowest_employee = employee
            lowest_value = salary

    return lowest_employee, lowest_value


def employees_above_average(employees):
    """Return employees whose salary is greater than the average."""
    average = average_salary(employees)
    above_average = []

    for employee, salary in employees.items():
        if salary > average:
            above_average.append(employee)

    return above_average


employees = {
    "Amit": 60000,
    "Priya": 75000,
    "Rahul": 50000,
    "Neha": 90000
}

print("Total payroll:", total_payroll(employees))
print("Average salary:", average_salary(employees))
print("Highest salary:", highest_salary(employees))
print("Lowest salary:", lowest_salary(employees))
print("Employees above average:", employees_above_average(employees))
```

## 7. Shopping Cart

Users can add products, remove products, view the cart, calculate the bill, and exit.

```python
def add_product_to_cart(cart):
    """Accept and add one product to the shopping cart."""
    product_name = input("Enter product name: ")
    price = float(input("Enter product price: "))
    quantity = int(input("Enter quantity: "))

    if price <= 0 or quantity <= 0:
        print("Price and quantity must be greater than zero")
        return

    cart[product_name] = {"price": price, "quantity": quantity}
    print("Product added")


def remove_product_from_cart(cart):
    """Remove a product from the shopping cart."""
    product_name = input("Enter product name to remove: ")

    if product_name in cart:
        del cart[product_name]
        print("Product removed")
    else:
        print("Product not found")


def view_cart(cart):
    """Display every product currently in the cart."""
    if len(cart) == 0:
        print("Cart is empty")
        return

    for product_name, details in cart.items():
        print(product_name, "->", details)


def calculate_bill(cart):
    """Return the total shopping-cart bill."""
    total = 0

    for details in cart.values():
        total = total + details["price"] * details["quantity"]

    return total


def shopping_cart_application():
    """Run the shopping-cart menu until Exit is selected."""
    cart = {}

    while True:
        choice = input("""
1. Add product
2. Remove product
3. View cart
4. Calculate bill
5. Exit
Enter choice (1-5): """).strip()

        if choice == "1":
            add_product_to_cart(cart)
        elif choice == "2":
            remove_product_from_cart(cart)
        elif choice == "3":
            view_cart(cart)
        elif choice == "4":
            print("Total bill:", calculate_bill(cart))
        elif choice == "5":
            print("Shopping cart closed")
            break
        else:
            print("Invalid choice")


shopping_cart_application()
```

## 8. Password Strength Checker

The function checks uppercase, lowercase, number, special character, and a minimum length of 8.

```python
def check_password_strength(password):
    """Return a meaningful password-strength result."""
    has_uppercase = False
    has_lowercase = False
    has_number = False
    has_special_character = False

    for character in password:
        if character.isupper():
            has_uppercase = True
        elif character.islower():
            has_lowercase = True
        elif character.isdigit():
            has_number = True
        else:
            has_special_character = True

    missing_requirements = []

    if len(password) < 8:
        missing_requirements.append("minimum 8 characters")
    if has_uppercase == False:
        missing_requirements.append("uppercase letter")
    if has_lowercase == False:
        missing_requirements.append("lowercase letter")
    if has_number == False:
        missing_requirements.append("number")
    if has_special_character == False:
        missing_requirements.append("special character")

    if len(missing_requirements) == 0:
        return "Strong password"

    return "Weak password. Missing: " + ", ".join(missing_requirements)


password = input("Enter a password: ")
print(check_password_strength(password))
```

## 9. Prime Number Analyzer

Enter two numbers representing a range. Functions find, count, sum, and display the largest prime.

```python
def is_prime(number):
    """Return True when a number is prime and False otherwise."""
    if number < 2:
        return False

    for divisor in range(2, number):
        if number % divisor == 0:
            return False

    return True


def find_prime_numbers(start, end):
    """Return all prime numbers in the inclusive range."""
    prime_numbers = []

    for number in range(start, end + 1):
        if is_prime(number):
            prime_numbers.append(number)

    return prime_numbers


def count_primes(prime_numbers):
    """Return the number of prime numbers."""
    count = 0

    for number in prime_numbers:
        count = count + 1

    return count


def calculate_prime_sum(prime_numbers):
    """Return the total of all prime numbers."""
    total = 0

    for number in prime_numbers:
        total = total + number

    return total


def largest_prime(prime_numbers):
    """Return the largest prime or None when no prime exists."""
    if len(prime_numbers) == 0:
        return None

    largest = prime_numbers[0]

    for number in prime_numbers[1:]:
        if number > largest:
            largest = number

    return largest


def prime_number_analyzer():
    """Accept a range and display the complete prime-number analysis."""
    start = int(input("Enter the start of the range: "))
    end = int(input("Enter the end of the range: "))

    if start > end:
        start, end = end, start

    prime_numbers = find_prime_numbers(start, end)
    print("Prime numbers:", prime_numbers)
    print("Prime count:", count_primes(prime_numbers))
    print("Prime sum:", calculate_prime_sum(prime_numbers))
    print("Largest prime:", largest_prime(prime_numbers))


prime_number_analyzer()
```

## 10. Expense Tracker

Users can add expenses, view expenses, calculate total, find the highest expense, and exit.

```python
def add_expense(expenses):
    """Accept and add one expense."""
    expense_name = input("Enter expense name: ")
    amount = float(input("Enter amount: "))

    if amount <= 0:
        print("Amount must be greater than zero")
        return

    expenses.append({"expense_name": expense_name, "amount": amount})
    print("Expense added")


def view_expenses(expenses):
    """Display every recorded expense."""
    if len(expenses) == 0:
        print("No expenses available")
        return

    for expense in expenses:
        print(expense["expense_name"], "->", expense["amount"])


def calculate_expense_total(expenses):
    """Return the total of all expenses."""
    total = 0

    for expense in expenses:
        total = total + expense["amount"]

    return total


def find_highest_expense(expenses):
    """Return the expense with the highest amount or None."""
    if len(expenses) == 0:
        return None

    highest = expenses[0]

    for expense in expenses[1:]:
        if expense["amount"] > highest["amount"]:
            highest = expense

    return highest


def expense_tracker():
    """Run the expense-tracker menu until Exit is selected."""
    expenses = []

    while True:
        choice = input("""
1. Add expense
2. View expenses
3. Calculate total
4. Find highest expense
5. Exit
Enter choice (1-5): """).strip()

        if choice == "1":
            add_expense(expenses)
        elif choice == "2":
            view_expenses(expenses)
        elif choice == "3":
            print("Total expense:", calculate_expense_total(expenses))
        elif choice == "4":
            print("Highest expense:", find_highest_expense(expenses))
        elif choice == "5":
            print("Expense tracker closed")
            break
        else:
            print("Invalid choice")


expense_tracker()
```

## 11. Mini Authentication System

Predefined credentials: username `admin`, password `python123`. The system allows a maximum of three attempts and supports logout and retry.

```python
PREDEFINED_USERNAME = "admin"
PREDEFINED_PASSWORD = "python123"
MAX_ATTEMPTS = 3


def login():
    """Allow a maximum of three login attempts and return the result."""
    for attempt in range(1, MAX_ATTEMPTS + 1):
        username = input("Username: ")
        password = input("Password: ")

        if username == PREDEFINED_USERNAME and password == PREDEFINED_PASSWORD:
            print("Login successful")
            return True

        print("Invalid username or password")

    print("Login failed after 3 attempts")
    return False


def logout():
    """Display a logout confirmation."""
    print("Logout successful")


def authentication_system():
    """Run login, logout, and retry logic."""
    while True:
        login_successful = login()

        if login_successful:
            input("Press Enter to logout: ")
            logout()

        retry = input("Do you want to try again? (yes/no): ").strip().lower()

        if retry != "yes":
            print("Authentication system closed")
            break


authentication_system()
```

## 12. Super30 Python Utility Application

Five utilities: Calculator, Palindrome checker, Prime checker, Factorial calculator, and Multiplication table.

```python
def calculator():
    """Run a basic calculation using two numbers and one operator."""
    number1 = float(input("Enter first number: "))
    operator = input("Enter operator (+, -, *, /): ")
    number2 = float(input("Enter second number: "))

    if operator == "+":
        print("Result:", number1 + number2)
    elif operator == "-":
        print("Result:", number1 - number2)
    elif operator == "*":
        print("Result:", number1 * number2)
    elif operator == "/":
        if number2 == 0:
            print("Cannot divide by zero")
        else:
            print("Result:", number1 / number2)
    else:
        print("Invalid operator")


def palindrome_checker():
    """Check whether an entered string is a palindrome."""
    text = input("Enter a string: ")
    reversed_text = ""

    for character in text:
        reversed_text = character + reversed_text

    if text == reversed_text:
        print("Palindrome")
    else:
        print("Not a palindrome")


def prime_checker():
    """Check whether an entered number is prime."""
    number = int(input("Enter an integer: "))

    if number < 2:
        print("Not prime")
        return

    for divisor in range(2, number):
        if number % divisor == 0:
            print("Not prime")
            return

    print("Prime")


def factorial_calculator():
    """Calculate the factorial of an entered non-negative integer."""
    number = int(input("Enter a non-negative integer: "))

    if number < 0:
        print("Factorial is not defined for negative numbers")
        return

    factorial = 1

    for value in range(1, number + 1):
        factorial = factorial * value

    print("Factorial:", factorial)


def multiplication_table():
    """Display the multiplication table of an entered number from 1 to 10."""
    number = int(input("Enter an integer: "))

    for multiplier in range(1, 11):
        print(number, "x", multiplier, "=", number * multiplier)


def super30_utility_application():
    """Run the five-utility menu until Exit is selected."""
    while True:
        choice = input("""
1. Calculator
2. Palindrome checker
3. Prime checker
4. Factorial calculator
5. Multiplication table
6. Exit
Enter choice (1-6): """).strip()

        if choice == "1":
            calculator()
        elif choice == "2":
            palindrome_checker()
        elif choice == "3":
            prime_checker()
        elif choice == "4":
            factorial_calculator()
        elif choice == "5":
            multiplication_table()
        elif choice == "6":
            print("Super30 utility application closed")
            break
        else:
            print("Invalid choice")


super30_utility_application()
```

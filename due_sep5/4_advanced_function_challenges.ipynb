## 1. Create a function using *args that accepts any number of values and returns their total.

```python
def total_values(*args):
    total = 0

    for value in args:
        total = total + value

    return total


print("Total:", total_values(10, 20, 30, 40))
```

## 2. Create a function using *args that returns the largest supplied number.

```python
def largest_value(*args):
    if len(args) == 0:
        return None

    largest = args[0]

    for value in args[1:]:
        if value > largest:
            largest = value

    return largest


print("Largest:", largest_value(12, 45, 7, 89, 34))
```

## 3. Create def create_profile(**kwargs)

that accepts dynamic user information and prints all provided attributes.

```python
def create_profile(**kwargs):
    for key, value in kwargs.items():
        print(key, "->", value)


create_profile(
    name="Rahul",
    age=22,
    course="Data Science",
    city="Bangalore"
)
```

## 4. Pass a Function as an Argument

```python
def add(number1, number2):
    return number1 + number2


def multiply(number1, number2):
    return number1 * number2


def calculate(operation, number1, number2):
    return operation(number1, number2)


print("Addition:", calculate(add, 10, 20))
print("Multiplication:", calculate(multiply, 10, 20))
```

## 6. Use lambda with map() to square

map(lambda x: expression, iterable)

```python
numbers = [1, 2, 3, 4, 5, 6]

squared_numbers = list(map(lambda number: number * number, numbers))

print("Original numbers:", numbers)
print("Squared numbers:", squared_numbers)
```

## 7. Use Lambda with `filter()`

filter() is used to keep only the items that satisfy a condition.

filter(function, iterable) 


filter(lambda x: condition, iterable) # With lambda

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda number: number % 2 == 0, numbers))

print("Original numbers:", numbers)
print("Even numbers:", even_numbers)
```

## 8. Recursive Factorial

Base case: `factorial(0)` and `factorial(1)` return `1`.

```python

def factorial(number):
    """Return the factorial of a non-negative integer."""
    print(f"Calling factorial({number})")

    if number == 0 or number == 1:
        print(f"factorial({number}) returns 1")
        return 1

    smaller_result = factorial(number - 1)

    result = number * smaller_result
    print(f"factorial({number}) returns {result}")

    return result


final_answer = factorial(5)
print("Final answer:", final_answer)
```

## 9. Recursive Sum from 1 to n

Base case: `recursive_sum(0)` returns `0`.

```python
def recursive_sum(number):
    if number < 0:
        return "Enter a non-negative number"

    if number == 0:
        return 0

    return number + recursive_sum(number - 1)


print("Sum from 1 to 10:", recursive_sum(10))
```

## 10. Recursive Fibonacci Sequence

The function calculates the Fibonacci number at a supplied position.

```python
def fibonacci(position):
    if position == 0:
        return 0

    if position == 1:
        return 1

    return fibonacci(position - 1) + fibonacci(position - 2)


print("First 10 Fibonacci numbers:")

for position in range(10):
    print(fibonacci(position), end=" ")
```

## 11. Local and Global Variable Scope

A global variable is available throughout the program. A local variable exists only inside its function.

```python
global_message = "I am a global variable"


def show_scope():
    local_message = "I am a local variable"

    print("Inside function:", global_message)
    print("Inside function:", local_message)


show_scope()
print("Outside function:", global_message)

# local_message cannot be used here because it is local to show_scope().
```

## 12. Mini Calculator

Each operation has a separate function, and `main()` controls the program.

Sample inputs: choice `1`, first number `10`, second number `20`.

```python
def add(number1, number2):
    return number1 + number2


def subtract(number1, number2):
    return number1 - number2


def multiply(number1, number2):
    return number1 * number2


def divide(number1, number2):
    if number2 == 0:
        return "Cannot divide by zero"

    return number1 / number2


def main():
    choice = input("""
Choose an operation:
1. Add
2. Subtract
3. Multiply
4. Divide
Enter choice (1-4): """).strip()

    if int(choice) >4 or int(choice) <1:
        result = "Invalid choice; choose a valid number"
        return result

    number1 = float(input("Enter the first number: "))
    number2 = float(input("Enter the second number: "))

    if choice == "1":
        result = add(number1, number2)
    elif choice == "2":
        result = subtract(number1, number2)
    elif choice == "3":
        result = multiply(number1, number2)
    elif choice == "4":
        result = divide(number1, number2)

    print("Result:", result)


main()
```

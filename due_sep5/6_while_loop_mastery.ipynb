# while Loop Mastery

## Question 1

Print numbers from 1 to 100 using while.

````python
# Initialization: begin with the first number.
number = 1

# Condition: continue while number is 100 or smaller.
while number <= 100:
    print(number, end=" ")
    # Update: increase number so the loop eventually stops.
    number += 1
````

## Question 2

Print numbers from 100 to 1.

````python
# Initialization: begin with 100.
number = 100

# Condition: continue while number is at least 1.
while number >= 1:
    print(number, end=" ")
    # Update: decrease number so the loop eventually stops.
    number -= 1
````

Print all even numbers between 1 and 100.

## Question 4

Calculate the sum of digits of a number.

**Example**

Input: 5832

Output: 18

````python
user_number = input("Enter a number: ")

user_number_list = [int(x) for x in user_number]

digit_sum = 0

while len(user_number_list) > 0:
    digit_sum = digit_sum + user_number_list.pop()


print("Sum of digits:", digit_sum)
````

## Question 7

Calculate factorial using while.

````python
number = int(input("Enter a non-negative integer: "))

if number < 0:
    print("Factorial is not defined for negative numbers")
else:
    # Initialization: factorial starts at 1 and counting begins at 1.
    factorial = 1
    counter = 1

    # Condition: multiply through the entered number.
    while counter <= number:
        factorial *= counter
        # Update: move to the next multiplier.
        counter += 1

    print("Factorial:", factorial)
````

## Question 8

Create a program that repeatedly asks for numbers and stops only when the user enters 0. Display the sum of all previously entered numbers.

````python
# Initialization: begin the sum at zero.
total = 0

# Condition: True keeps asking because the number of entries is unknown.
while True:
    number = float(input("Enter a number (0 to stop): "))

    # Termination: entering 0 exits the loop.
    if number == 0:
        break

    # Update: add the latest non-zero number to the running total.
    total += number

print("Sum of all previously entered numbers:", total)
````

## Question 9

Create a password checker that keeps asking for the password until the correct password is entered.

````python
correct_password = "python123"

# Initialization: ask for the first password.
entered_password = input("Enter the password: " )

# Condition: continue while the entered password is incorrect.
while entered_password != correct_password:
    print("Incorrect password. Try again.")
    # Update: ask again; a correct value makes the condition false.
    entered_password = input("Enter the password: " )

print("Correct password. Access granted.")
````

## Question 10

Create a guessing game where the secret number is predefined and the user keeps guessing until correct.

````python
secret_number = 7

# Initialization: None cannot equal the integer secret number.
guess = None

# Condition: continue until the guess equals the secret number.
while guess != secret_number:
    # Update: collect a new guess during every iteration.
    guess = int(input("Guess the secret number: "))

    if guess < secret_number:
        print("Too low")
    elif guess > secret_number:
        print("Too high")

print("Correct guess!")
````

## Question 11

Build a menu-driven calculator

```text
1
Add

2
Subtract

3
Multiply

4
Divide

5
Exit
```

The menu should continue until the user chooses Exit.

````python
# Initialization: use a value other than 5 so the menu opens.
choice = 0

# Condition: continue until the user chooses 5 (Exit).
while choice != 5:
    print("\n1")
    print("Add")
    print("2")
    print("Subtract")
    print("3")
    print("Multiply")
    print("4")
    print("Divide")
    print("5")
    print("Exit")

    # Update/termination: a new choice controls the next iteration.
    choice = int(input("Enter your choice: "))

    if choice == 5:
        print("Calculator closed.")
        break

    if choice not in [1, 2, 3, 4]:
        print("Invalid choice")
        continue

    first_number = float(input("Enter first number: "))
    second_number = float(input("Enter second number: "))

    if choice == 1:
        print("Result:", first_number + second_number)
    elif choice == 2:
        print("Result:", first_number - second_number)
    elif choice == 3:
        print("Result:", first_number * second_number)
    elif second_number == 0:
        print("Cannot divide by zero")
    else:
        print("Result:", first_number / second_number)
````

## Question 12

Create an ATM menu using while where the application continues running until the user explicitly chooses Exit.

````python
balance = 1000.0

# Initialization: use a value other than 4 so the ATM menu opens.
choice = 0

# Condition: continue until the user explicitly chooses 4 (Exit).
while choice != 4:
    print("\nATM Menu")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    # Update/termination: a new choice controls the next iteration.
    choice = int(input("Enter your choice: "))

    if choice == 1:
        print("Current balance:", balance)
    elif choice == 2:
        amount = float(input("Enter deposit amount: "))
        if amount > 0:
            balance += amount
            print("Deposit successful. Total balance is:",balance )
        else:
            print("Invalid amount")
    elif choice == 3:
        amount = float(input("Enter withdrawal amount: "))
        if amount <= 0:
            print("Invalid amount")
        elif amount > balance:
            print("Insufficient balance")
        else:
            balance -= amount
            print("Withdrawal successful. Total balance is:",balance )
    elif choice == 4:
        print("Thank you for using the ATM.")
    else:
        print("Invalid choice")
````

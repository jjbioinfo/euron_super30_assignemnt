# Intermediate Loops & Loop Control

## Question 1

Print numbers from 1–100, but skip numbers divisible by 5 using `continue`.

```python
# Question 1
for number in range(1, 101):
    if number % 5 == 0:
        continue
    print(number, end=" ")
```

## Question 2

Iterate from 1–100 and stop at the first number divisible by both 7 and 11.

```python
# Question 2
for number in range(1, 101):
    if number % 7 == 0 and number % 11 == 0:
        print("First matching number:", number)
        break
```

## Question 3

Search for a user-provided number in a list using `for-else`.

```python
# Question 3
numbers = [10, 20, 30, 40, 50, 60]
search_number = int(input("Enter a number to search: "))

for number in numbers:
    if number == search_number:
        print("Number Found", number)
        break
else:
    print("Number Not Found")
```

## Question 4

Use `enumerate()` to display the names with numbering starting from 1.

enumerate() gives you two things while looping:
1. The position number
2. The item itself

```python
# Question 4
names = ["Aman", "Ravi", "Sudhanshu", "Priya", "Anjali"]

for position, name in enumerate(names, start=1):
    print(position, name)
```

## Question 5

Print an increasing star pattern.

```python
# Question 5
for row in range(1, 6):
    print("*" * row)
```

## Question 6

Print a decreasing star pattern.

```python
# Question 6
for row in range(5, 0, -1):
    print("*" * row)
```

## Question 7

Generate multiplication tables from 1 to 10 using nested loops.

```python
# Question 7
for table_number in range(1, 11):
    print(f"\nMultiplication table of {table_number}")
    for multiplier in range(1, 11):
        product = table_number * multiplier
        print(f"{table_number} x {multiplier} = {product}")
```

## Question 9

Given a list containing duplicate elements, create another list containing only unique elements without using set().

```python
# Question 9
duplicate_values = [1, 2, 2, 3, 1, 4, 3, 5, 5]
unique_values = []

for value in duplicate_values:
    if value not in unique_values:
        unique_values.append(value)

print("Original list:", duplicate_values)
print("Unique list:", unique_values)
```

## Question 10

Count positive numbers, negative numbers, and zeros.

```python
# Question 10
numbers = [10, -4, 8, -2, 0, 15, -9, 21]
positive_count = 0
negative_count = 0
zero_count = 0

for number in numbers:
    if number > 0:
        positive_count += 1
    elif number < 0:
        negative_count += 1
    else:
        zero_count += 1

print("Positive numbers:", positive_count)
print("Negative numbers:", negative_count)
print("Zeros:", zero_count)
```

## Question 11

Determine whether a user-provided number is prime using a loop.

```python
# Question 11
number = int(input("Enter a number to check: "))

if number < 2:
    print(number, "is not a prime number")
else:
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            print(number, "is not a prime number")
            break
    else:
        print(number, "is a prime number")
```

## Question 12

Print all prime numbers from 1–100.

```python
# Question 12
for number in range(2, 101):
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            break
    else:
        print(number, end=" ")
```

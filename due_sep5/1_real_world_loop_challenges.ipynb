## calculate total transaction value without using sum().

From the same list, find the highest and lowest transaction without max() and min().

```python
transactions = [1200, 450, 800, 1500, 2300, 700, 100]

total = 0
highest = transactions[0]
lowest = transactions[0]

for transaction in transactions:
    total = total + transaction

    if transaction > highest:
        highest = transaction

    if transaction < lowest:
        lowest = transaction

print("Transactions:", transactions)
print("Total transaction value:", total)
print("Highest transaction:", highest)
print("Lowest transaction:", lowest)
```

## 2. find the average temperature.


```python
temperatures = [32, 35, 28, 40, 38, 31, 42]

total_temperature = 0

for temperature in temperatures:
    total_temperature = total_temperature + temperature

average_temperature = total_temperature / len(temperatures)

print("Temperatures:", temperatures)
print("Average temperature:", average_temperature)
```

## 4. Student Marks Analysis

count how many students scored : 90+, 75–89, 50–74, below 50

```python
marks = [78, 92, 45, 67, 88, 53, 99]

score_90_plus = 0
score_75_to_89 = 0
score_50_to_74 = 0
score_below_50 = 0

for mark in marks:
    if mark >= 90:
        score_90_plus = score_90_plus + 1
    elif mark >= 75:
        score_75_to_89 = score_75_to_89 + 1
    elif mark >= 50:
        score_50_to_74 = score_50_to_74 + 1
    else:
        score_below_50 = score_below_50 + 1

print("Marks:", marks)
print("90+:", score_90_plus)
print("75-89:", score_75_to_89)
print("50-74:", score_50_to_74)
print("Below 50:", score_below_50)
```

## 5. Login System — Maximum 3 Attempts

Test password: `python123`

```python
correct_password = "python123"
login_successful = False

for attempt in range(1, 4):
    password = input(f"Enter password - attempt {attempt}/3: ")

    if password == correct_password:
        print("Login successful")
        login_successful = True
        break
    else:
        print("Incorrect password")

if login_successful == False:
    print("Login failed after 3 attempts")
```

## 8. Count the frequency of every character in a string without using Counter.

Example input: `banana`

```python
text = input("Enter a string: ")
frequency = {}

for character in text:
    if character in frequency:
        frequency[character] = frequency[character] + 1
    else:
        frequency[character] = 1

print("user provided text:", text)
for character, count in frequency.items():
    print(character, "->", count)
```

## 9. Second-Largest Number Without `sort()`

```python
numbers = [42, 17, 68, 55, 68, 31, 60]

largest = numbers[0]
second_largest = None

for number in numbers[1:]:
    if number > largest:
        second_largest = largest
        largest = number
    elif number != largest:
        if second_largest is None or number > second_largest:
            second_largest = number

print("Numbers:", numbers)
print("Second-largest number:", second_largest)
```

## 10. Check whether a string is a palindrome using loops.

```python
text = input("Enter a string: ")
reversed_text = ""

for character in text:
    reversed_text = character + reversed_text
    #print(reversed_text)

if text == reversed_text:
    print(text, "is a palindrome")
else:
    print(text, "is not a palindrome")
```

## 12. Basic ATM Simulation

This uses one simple `while` loop. No functions are used.

```python
balance = 10000.0

while True:
    choice = input("""
Choose an option:
1. Check balance
2. Deposit money
3. Withdraw money
4. Exit
Enter choice (1-4): """).strip()
    if choice == "1":
        print("Balance: ₹", balance)

    elif choice == "2":
        amount = float(input("Enter deposit amount: ₹"))

        if amount > 0:
            balance = balance + amount
            print("Deposit successful. Balance: ₹", balance)
        else:
            print("Deposit amount must be greater than zero")

    elif choice == "3":
        amount = float(input("Enter withdrawal amount: ₹"))

        if amount <= 0:
            print("Withdrawal amount must be greater than zero")
        elif amount > balance:
            print("Insufficient balance")
        else:
            balance = balance - amount
            print("Withdrawal successful. Balance: ₹", balance)

    elif choice == "4":
        print("Thank you")
        break

    else:
        print("Invalid choice")
```

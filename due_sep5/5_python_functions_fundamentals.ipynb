## 1. Create `def add(a, b)` That Returns the Addition of Two Numbers

```python
def add(a, b):
    """Return the addition of two numbers."""
    return a + b


result = add(10, 20)
print("Addition:", result)
```

## 2. Create Functions for Addition, Subtraction, Multiplication, and Division

```python
def addition(a, b):
    """Return the addition of two numbers."""
    return a + b


def subtraction(a, b):
    """Return the difference between two numbers."""
    return a - b


def multiplication(a, b):
    """Return the product of two numbers."""
    return a * b


def division(a, b):
    """Return the division of two numbers or a message when dividing by zero."""
    if b == 0:
        return "Cannot divide by zero"

    return a / b


print("Addition:", addition(20, 5))
print("Subtraction:", subtraction(20, 5))
print("Multiplication:", multiplication(20, 5))
print("Division:", division(20, 5))
```

## 3. Create a Function That Determines Whether a Number Is Even or Odd

```python
def even_or_odd(number):
    """Return whether a number is even or odd."""
    if number % 2 == 0:
        return "Even"

    return "Odd"


print("10 is:", even_or_odd(10))
print("7 is:", even_or_odd(7))
```

## 4. Create a Function That Returns the Largest of Three Numbers Without Using `max()`

```python
def largest_of_three(number1, number2, number3):
    """Return the largest of three supplied numbers without using max()."""
    largest = number1

    if number2 > largest:
        largest = number2

    if number3 > largest:
        largest = number3

    return largest


print("Largest number:", largest_of_three(25, 80, 45))
```

## 5. Create a Function That Calculates Factorial

```python
def factorial(number):
    """Return the factorial of a non-negative integer."""
    if number < 0:
        return "Factorial is not defined for negative numbers"

    result = 1

    for value in range(1, number + 1):
        result = result * value

    return result


print("Factorial of 5:", factorial(5))
print("Factorial of 0:", factorial(0))
```

## 6. Create a Function That Checks Whether a Number Is Prime

```python
def is_prime(number):
    """Return True when a number is prime and False otherwise."""
    if number < 2:
        return False

    for divisor in range(2, number):
        if number % divisor == 0:
            return False

    return True


print("11 is prime:", is_prime(11))
print("12 is prime:", is_prime(12))
```

## 7. Create `def calculate_discount(price, discount=10)`

The default discount is 10%.

```python
def calculate_discount(price, discount=10):
    """Return the final price after applying a percentage discount."""
    discount_amount = price * discount / 100
    final_price = price - discount_amount
    return final_price


print("Default 10% discount:", calculate_discount(1000))
print("Custom 20% discount:", calculate_discount(1000, 20))
```

```python
print("Custom 20% discount:", calculate_discount(price=1000, discount=20))
```

## 8. Create a Function That Accepts a List and Returns Its Sum Without Using `sum()`

```python
def list_total(numbers):
    """Return the total of all numbers in a list without using sum()."""
    total = 0

    for number in numbers:
        total = total + number

    return total


values = [10, 20, 30, 40]
print("List total:", list_total(values))
```

## 9. Create a Function That Accepts a String and Returns the Number of Vowels

```python
def count_vowels(text):
    """Return the number of vowels in a string."""
    vowel_count = 0

    for character in text.lower():
        if character in "aeiou":
            vowel_count = vowel_count + 1

    return vowel_count


print("Vowels in education:", count_vowels("education"))
```

## 10. Create a Function That Accepts a String and Determines Whether It Is a Palindrome

```python
def is_palindrome(text):
    """Return True when a string is a palindrome and False otherwise."""
    reversed_text = ""

    for character in text:
        reversed_text = character + reversed_text

    return text == reversed_text


print("madam is a palindrome:", is_palindrome("madam"))
print("python is a palindrome:", is_palindrome("python"))
```

## 11. Create a Function That Accepts Name, Age, and Course and Returns a Formatted Student Profile

Call it using both positional and keyword arguments.

```python
def student_profile(name, age, course):
    """Return a formatted student profile containing name, age, and course."""
    return f"Name: {name}, Age: {age}, Course: {course}"


positional_profile = student_profile("Rahul", 22, "Data Science")
keyword_profile = student_profile(
    name="Priya",
    age=21,
    course="Python"
)

print("Positional arguments:", positional_profile)
print("Keyword arguments:", keyword_profile)
```

## 12. Demonstrate the Difference Between a Function That Uses `print()` and a Function That Uses `return`

A printed value is displayed but cannot be reused as a result. A returned value can be saved, passed elsewhere, or used in another calculation.

```python
def add_with_print(a, b):
    """Print the addition of two numbers without returning the result."""
    print("Printed inside function:", a + b)


def add_with_return(a, b):
    """Return the addition of two numbers for later use."""
    return a + b


printed_result = add_with_print(10, 20)
returned_result = add_with_return(10, 20)

print("Value received from print function:", printed_result)
print("Value received from return function:", returned_result)
print("Returned value multiplied by 2:", returned_result * 2)
```

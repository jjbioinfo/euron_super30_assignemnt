# Create a Student Marks List

Run the code cell below to complete all operations in the assignment.


```python
marks = [78, 85, 90, 67, 88, 92, 76]

# Print the complete list
print("Complete list:", marks)

# Print the first and last elements
print("First element:", marks[0])
print("Last element:", marks[-1])

# Print elements from index 2 to index 5
print("Elements from index 2 to 5:", marks[2:6])

# Find the number of elements, maximum, minimum, and total
print("Number of elements:", len(marks))
print("Maximum marks:", max(marks))
print("Minimum marks:", min(marks))
print("Total marks:", sum(marks))

# Sort marks in ascending order
ascending_marks = marks.copy()
ascending_marks.sort()
print("Ascending order:", ascending_marks)

# Sort marks in descending order
descending_marks = marks.copy()
descending_marks.sort(reverse=True)
print("Descending order:", descending_marks)

# Add 95
marks.append(95)
print("After adding 95:", marks)

# Add 81 and 84
marks.extend([81, 84])
print("After adding [81, 84]:", marks)

# Remove 67
marks.remove(67)
print("After removing 67:", marks)

# Count 90 and find the index of 88
print("Number of times 90 occurs:", marks.count(90))
print("Index of 88:", marks.index(88))
```
